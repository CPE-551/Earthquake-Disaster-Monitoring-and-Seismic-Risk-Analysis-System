```python
#main.ipynb
"""
Project: Earthquake Disaster Monitoring and Seismic Risk Analysis System

File:
    main.ipynb

Description:
    Main execution notebook for the earthquake risk analysis system.

Function:
    - Loads USGS earthquake dataset.
    - Converts records into Earthquake objects.
    - Performs seismic analysis.
    - Generates risk reports.
    - Exports results.
    - Creates visualization charts.

Libraries:
    Pandas:
        Used inside data_loader.py for CSV processing.

    NumPy:
        Used inside analyzer.py for statistical calculations.

    Matplotlib:
        Used inside visualization.py for earthquake charts.

"""


# Import data loading function.
from data_loader import load_earthquake_data


# Import earthquake analyzer class.
from analyzer import EarthquakeAnalyzer


# Import visualization functions.
from visualization import (
    plot_magnitude_distribution,
    plot_frequency_distribution,
    plot_depth_vs_magnitude
)


# Import report exporting functions.
from report import (
    export_risk_report,
    export_statistics
)


# Import generator function.
from generator import earthquake_generator


def main():
    """
    Description:
        Main function that executes the complete earthquake analysis system.

    Function:
        1. Loads earthquake data from USGS dataset.
        2. Creates Earthquake objects.
        3. Stores objects inside EarthquakeAnalyzer.
        4. Calculates statistics.
        5. Generates risk assessment report.
        6. Exports reports and charts.

    Returns:
        None
    """

    try:
        
        # STEP 1:
        # Load earthquake records from USGS CSV dataset.

        earthquakes = load_earthquake_data(
            "data/earthquakes.csv"
        )


        print(
            f"Loaded {len(earthquakes)} earthquake records."
        )

    except FileNotFoundError:

        print(
            "Dataset file was not found."
        )

        return

    except Exception as error:

        print(
            f"Dataset loading error: {error}"
        )

        return


    # STEP 2:
    # Create EarthquakeAnalyzer object.
    # Composition relationship:
    # EarthquakeAnalyzer contains Earthquake objects.
   

    analyzer = EarthquakeAnalyzer()

   
    # STEP 3:
    # Process earthquake objects using generator.
    # Generator:
    # Processes one earthquake at a time.
    # Loop requirement:
    # Uses for loop.

    for earthquake in earthquake_generator(earthquakes):

        analyzer.add_earthquake(
            earthquake
        )


    # STEP 4:
    # Statistical Analysis

    average_magnitude = (
        analyzer.calculate_average_magnitude()
    )


    strongest_event = (
        analyzer.find_strongest_earthquake()
    )


    frequency_results = (
        analyzer.analyze_frequency()
    )

    print("\n===")
    print("Earthquake Analysis Summary")
    print("====")

    print(
        f"Total Earthquakes: {len(analyzer)}"
    )


    print(
        f"Average Magnitude: {average_magnitude:.2f}"
    )


    print(
        "\nStrongest Earthquake:"
    )


    print(
        strongest_event
    )


    # STEP 5:
    # List comprehension example.
    # Finds earthquakes with magnitude >= 6.
 
    major_events = [
        earthquake
        for earthquake in analyzer.earthquakes
        if earthquake.magnitude >= 6
    ]

    print(
        f"\nHigh Risk Earthquakes: {len(major_events)}"
    )


    # STEP 6:
    # Generate Risk Assessment Report
 
    risk_report = (
        analyzer.generate_risk_report()
    )

    print("\nRisk Report")

    for key, value in risk_report.items():

        print(
            f"{key}: {value}"
        )


    # STEP 7:
    # Export Results
    # Output:
    # - Text report
    # - CSV statistics
    
    export_risk_report(
        risk_report,
        "earthquake_report.txt"
    )

    export_statistics(
        risk_report,
        "earthquake_statistics.csv"
    )

    print(
        "\nReports exported successfully."
    )


    # STEP 8:
    # Generate Visualization Charts
    # Output:
    # - Magnitude distribution chart
    # - Frequency chart
    # - Depth vs magnitude chart
  
    plot_magnitude_distribution(
        analyzer.earthquakes
    )


    plot_frequency_distribution(
        frequency_results
    )


    plot_depth_vs_magnitude(
        analyzer.earthquakes
    )


    print(
        "\nVisualization completed."
    )


# Program Entry Point
# Allows notebook/script execution while keeping modules reusable.

if __name__ == "__main__":

    main()
```
