```python
# main.ipynb
"""
Project: Earthquake Disaster Monitoring and Seismic Risk Analysis System

File:
    main.ipynb

Description:
    Main execution script for the earthquake risk analysis system.

Function:
    - Loads USGS earthquake dataset.
    - Converts records into Earthquake objects.
    - Performs seismic analysis.
    - Generates risk reports.
    - Exports results.
    - Creates visualization charts.

Libraries:
    Pandas:
        Used inside data_loader.py for CSV processing.

    NumPy:
        Used inside analyzer.py for statistical calculations.

    Matplotlib:
        Used inside visualization.py for earthquake charts.

"""
# Import data loading and export report functions, earthquake analyzer class
import os
from data_loader import load_earthquake_data, iter_earthquake_records
from analyzer import EarthquakeAnalyzer
from report import export_risk_report

from visualization import plot_magnitude_distribution, plot_frequency_distribution, plot_depth_vs_magnitude






def main():
    """
    Description:
        Main function that executes the complete earthquake analysis system.

    Function:
        1. Loads earthquake data from USGS dataset.
        2. Creates Earthquake objects.
        3. Stores objects inside EarthquakeAnalyzer.
        4. Calculates statistics.
        5. Generates risk assessment report.
        6. Exports reports and charts.

    Returns:
        None
    """

    file_path = "data/earthquakes.csv"
    # Earthquake risk report file address export
    output_file = "outputs/earthquake_risk_report.csv"

    output_directory = os.path.dirname(output_file) 

    if output_directory: #TC 
        os.makedirs(output_directory, exist_ok=True) 

    try:
        earthquakes = load_earthquake_data(file_path) # Earthquake object list is loaded and created

        print("Earthquake data loaded successfully.")
        print(f"Number of earthquake records: {len(earthquakes)}\n")

    except(FileNotFoundError, ValueError, OSError) as error: # Exception handling for missing dataset file.
        print("Data loading error:", error)
        return # Return to Main if no earthquake records were found

    
    # Initiate analyzer and populate data
    analyzer = EarthquakeAnalyzer()

    for earthquake in earthquakes:
        analyzer.add_earthquake(earthquake)

    print("EarthquakeAnalyzer created sucessfully.")
    print(f"Earthquakes added to analyzer: {len(analyzer.earthquakes)}\n")

    # Data is analyzed statistically
    average_magnitude = analyzer.calculate_average_magnitude()
    print(f"Average Earthquake Magnitude: {average_magnitude:.2f}")  

    strongest_event = analyzer.find_strongest_earthquake()
    print("\nStrongest Earthquake:")
    print(strongest_event)
    

    # Generate the earthquake risk report
    report = analyzer.generate_risk_report()
    print("Earthquake Risk Report:")
    print(report)

    try:
        export_risk_report(report, output_file)

        print("\nRisk report has been created sucessfully.")
        print(f"Report saved to the following address: {output_file}")
    
    except OSError as error:
        print("Report export error:", error)       


    plot_magnitude_distribution(analyzer.earthquakes,save_path="outputs/magnitude_distribution.png") # Visiualize the distribution of earthquake magnitudes across the dataset, save graph to outputs folder
    

    plot_frequency_distribution(report["Frequency"],save_path="outputs/frequency_distribution.png") # Visualize how many earthquakes fall into each severity category, save graph to outputs folder


    plot_depth_vs_magnitude(analyzer.earthquakes, save_path="outputs/depth_vs_magnitude.png") # Visualize the relationship between earthquake depth and magnitude, save graph to outputs folder


    # Call generator function that consumes the Earthquake data and stops as soon as it detects a Strong (6.0 - 6.9), Major (7.0 - 7.9), or Great (8.0 and higher) earthquake
    for earthquake in iter_earthquake_records(file_path):
        if earthquake.magnitude >= 6.0:
            print("Earthquake greater than 6.0 detected!")
            print(earthquake)      
            break
    else:
        print("All earthquakes in this dataset are moderate or lower in severity")
    
if __name__ == "__main__":
    main()
```
